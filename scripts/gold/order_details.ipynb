```python
%sql
create schema if not exists e_comm.gold
```

```python
spark.sql("create table if not exists e_comm.gold.order_details (order_id string, customer_id string, order_item_id int, seller_id string, price float, order_status string)")
```

```python
from pyspark.sql.functions import col
df_cust = spark.read.table("e_comm.silver.customers")
df_orders = spark.read.table("e_comm.silver.orders")
df_order_items = spark.read.table("e_comm.silver.order_items")

df = df_orders.alias("o").join(df_order_items.alias("i"), col('o.order_id') == col('i.order_id'), "inner").select(col('o.order_id'),col('o.customer_id'),col('i.order_item_id'),col('i.seller_id'),col('i.price'),col('o.order_status'),)

df.write.format("delta").mode("overwrite").saveAsTable("e_comm.gold.order_details")
```
