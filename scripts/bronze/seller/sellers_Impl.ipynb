```python
spark.sql("USE CATALOG e_comm")
spark.sql("USE SCHEMA bronze")
```

```python
from pyspark.sql.functions import lit, current_timestamp
import requests
import pandas as pd
from io import StringIO
url = "https://raw.githubusercontent.com/deepakmali17/E_commerce_dataplatform/refs/heads/main/datasets/sellers.csv"

response = requests.get(url)

response.raise_for_status()
pdf  = pd.read_csv(StringIO(response.text))
df = spark.createDataFrame(pdf)
df = df.withColumn("merge_flag", lit(False)).withColumn("ingestion_ts", current_timestamp())
display(df)
```

```python
df.write.format("delta").mode("append").option("mergeSchema" , True).saveAsTable("bronze.sellers")
```
