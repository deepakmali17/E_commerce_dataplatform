```python
from pyspark.sql.functions import *
import requests
import pandas as pd
from io import StringIO
url = "https://raw.githubusercontent.com/deepakmali17/E_commerce_dataplatform/refs/heads/main/datasets/products.csv"

response = requests.get(url)
response.raise_for_status()

pdf = pd.read_csv(StringIO(response.text))
df = spark.createDataFrame(pdf)

df = df.withColumn("merge_flag", lit(False))

```

```python
df = df.write.format("delta").mode("append").option("mergeSchema", "True").save("/Volumes/e_comm/bronze/volume/products/")
```
